# Load CSV into Dataframe
Alternatively, the LGS IT team also dumped the transactional data into a [CSV file](https://raw.githubusercontent.com/jarviscanada/jarvis_data_eng_demo/feature/data/python_data_wrangling/data/online_retail_II.csv). However, the CSV header (column names) doesn't follow the snakecase or camelcase naming convention (e.g. `Customer ID` instead of `customer_id` or `CustomerID`). As a result, you will need to use Pandas to clean up the data before doing any analytics. In addition, unlike the PSQL scheme, CSV files do not have data types associated. Therefore, you will need to cast/convert certain columns into correct data types (e.g. DateTime, numbers, etc..)

**Data Preperation**

- Read the `data/online_retail_II.csv` file into a DataFrame
- Rename all columns to upper camelcase or snakecase
- Convert/cast all columns to the appropriate data types (e.g. datetime)

```python
retail_df = pd.read_csv('data/online_retail_II.csv')

# Rename columns to snake_case
retail_df.rename(columns={
    'Invoice': 'invoice_no',
    'StockCode': 'stock_code',
    'Description': 'description',
    'Quantity': 'quantity',
    'InvoiceDate': 'invoice_date',
    'Price': 'unit_price',
    'Customer ID': 'customer_id',
    'Country': 'country'
}, inplace=True)

# Cast types
retail_df['invoice_date'] = pd.to_datetime(retail_df['invoice_date'], errors='coerce')
retail_df['customer_id'] = pd.to_numeric(retail_df['customer_id'], errors='coerce')
retail_df['total_amount'] = retail_df['quantity'] * retail_df['unit_price']

retail_df.head()
```

# Total Invoice Amount Distribution

```python
# Calculate the invoice amount
invoice_amount = retail_df.groupby('invoice_no')['total_amount'].sum().reset_index()
invoice_amount.columns = ['invoice_no', 'invoice_amount']

#Remove negative/cancelled invoices
invoice_amount = invoice_amount[invoice_amount['invoice_amount']>0]
min_val = invoice_amount['invoice_amount'].min()
max_val = invoice_amount['invoice_amount'].max()
mean_val = invoice_amount['invoice_amount'].mean()
median_val = invoice_amount['invoice_amount'].median()
mode_val = invoice_amount['invoice_amount'].mode()[0]

# Stats
print("Minimum:", round(min_val, 2))
print("Mean:", round(mean_val, 2))
print("Median:", round(median_val, 2))
print("Mode:", round(mode_val, 2))
print("Maximum:", round(max_val, 2))

# Full distribution
fig, (ax1, ax2) = plt.subplots(
    2, 
    1, 
    figsize=(14, 8))

# Histogram
sns.histplot(
    invoice_amount['invoice_amount'],
    bins=100,
    color='steelblue',
    ax=ax1 )

#Statistical lines
ax1.set_title('Invoice Amount Distribution')
ax1.set_xlabel('Invoice Amount (£)')
ax1.set_ylabel('Frequency')
ax1.set_xlim(0, 175000)
ax1.hist(invoice_amount['invoice_amount'],
         bins=10,
         color='steelblue',
         edgecolor='black')

# Add lines for the statistics values
ax1.axvline(x=min_val, color='gray', linestyle='dashed', linewidth=2)
ax1.axvline(x=mean_val, color='red', linestyle='dashed', linewidth=2)
ax1.axvline(x=median_val, color='cyan', linestyle='dashed', linewidth=2)
ax1.axvline(x=mode_val, color='yellow', linestyle='dashed', linewidth=2)
ax1.axvline(x=max_val, color='gray', linestyle='dashed', linewidth=2)

# Boxplot
ax2.boxplot(invoice_amount['invoice_amount'],
            vert=False,
            widths=0.6) 
ax2.set_title('Invoice Amount Boxplot')
ax2.set_xlabel('Invoice Amount (£)')

plt.tight_layout()
plt.show()
```

### ***Invoice Amounts Across the First 85 Quantiles***

```python
# Keep only the first 85 quantiles
q85 = invoice_amount['invoice_amount'].quantile(0.85)
invoice_amount_85 = invoice_amount[invoice_amount['invoice_amount'] <= q85]

# Stats
min_val = invoice_amount_85['invoice_amount'].min()
max_val = invoice_amount_85['invoice_amount'].max()
mean_val = invoice_amount_85['invoice_amount'].mean()
median_val = invoice_amount_85['invoice_amount'].median()
mode_val = invoice_amount_85['invoice_amount'].mode()[0]

print("Minimum:", round(min_val, 2))
print("Mean:", round(mean_val, 2))
print("Median:", round(median_val, 2))
print("Mode:", round(mode_val, 2))
print("Maximum:", round(max_val, 2))

# Distribution graph
fig, (ax1, ax2) = plt.subplots(
    2,
    1,
    figsize=(14, 7),
    gridspec_kw={'height_ratios': [3, 1]}
)

# Histogram
ax1.hist(invoice_amount_85['invoice_amount'],
         bins=12,
         color='#3b77af')

# Reference lines
ax1.axvline(min_val, color='gray', linestyle='--', linewidth=2)
ax1.axvline(mode_val, color='yellow', linestyle='--', linewidth=2)
ax1.axvline(mean_val, color='red', linestyle='--', linewidth=2)
ax1.axvline(median_val, color='cyan', linestyle='--', linewidth=2)
ax1.axvline(max_val, color='gray', linestyle='--', linewidth=2)

ax1.set_title('Data Distribution')
ax1.set_ylabel('Frequency')

# Boxplot
ax2.boxplot(invoice_amount_85['invoice_amount'],
            vert=False,
            widths=0.5)

plt.tight_layout()
plt.show()
```

# Monthly Placed and Canceled Orders

```python
# Cancelled orders start with 'C'
retail_df['is_cancelled'] = retail_df['invoice_no'].str.startswith('C')
retail_df['yyyymm'] = retail_df['invoice_date'].dt.year * 100 + retail_df['invoice_date'].dt.month

monthly = retail_df.groupby('yyyymm')['invoice_no'].nunique().reset_index()
monthly.columns = ['yyyymm', 'total_orders']

cancelled = retail_df[retail_df['is_cancelled']].groupby('yyyymm')['invoice_no'].nunique().reset_index()
cancelled.columns = ['yyyymm', 'cancelled_orders']

monthly = monthly.merge(cancelled, on='yyyymm', how='left').fillna(0)
monthly['placed_orders'] = monthly['total_orders'] - 2 * monthly['cancelled_orders']

monthly.plot(x='yyyymm', y=['placed_orders', 'cancelled_orders'], kind='bar', figsize=(14, 5))
plt.title('Monthly Placed and Cancelled Orders')
plt.show()
```

# New and Existing Users



---
**Please remove this insturction cell after you are done with coding**

- Plot a diagram to show new and exiting user for each month.
- A user is identified as a new user when he/she makes the first purchase
- A user is identified as an existing user when he/she made purchases in the past
- hints:
  - find out the first purchase year-month for each user and then join this data with the transactional data to help you identified new/exiting users

![](https://i.imgur.com/nWjnrpr.jpg)

---

```python
# Create a copy to work with
users_df = retail_df.copy()

# Add yyyymm column
users_df['yyyymm'] = users_df['invoice_date'].dt.year * 100 + users_df['invoice_date'].dt.month

# Find each user's first purchase month
first_purchase = users_df.groupby('customer_id')['yyyymm'].min().reset_index()
first_purchase.columns = ['customer_id', 'first_purchase_month']

# Merge and flag user type
users_df = users_df.merge(first_purchase, on='customer_id', how='left')
users_df['user_type'] = users_df.apply(
    lambda x: 'New' if x['yyyymm'] == x['first_purchase_month'] else 'Existing', axis=1
)

# Plot
user_type_monthly = users_df.groupby(['yyyymm', 'user_type'])['customer_id'].nunique().unstack().fillna(0)

user_type_monthly.plot(kind='bar', figsize=(14, 5))
plt.title('New vs Existing Users per Month')
plt.xlabel('Year Month')
plt.ylabel('# of Users')
plt.show()
```

## Finding RFM

RFM is a method used for analyzing customer value. It is commonly used in database marketing and direct marketing and has received particular attention in the retail and professional services industries. ([wikipedia](https://en.wikipedia.org/wiki/RFM_(market_research)))

Optional Reading: [Making Your Database Pay Off Using Recency Frequency and Monetary Analysis](http://www.dbmarketing.com/2010/03/making-your-database-pay-off-using-recency-frequency-and-monetary-analysis/)


RFM stands for three dimensions:

- Recency – How recently did the customer purchase?

- Frequency – How often do they purchase?

- Monetary Value – How much do they spend?

Note: To simplify the problem, let's keep all placed and canceled orders.


**Sample RFM table**

![](https://i.imgur.com/sXFIg6u.jpg)

```python
import datetime as dt

snapshot_date = pd.Timestamp.today()

rfm = retail_df.groupby('customer_id').agg(
    Recency=('invoice_date', lambda x: (snapshot_date - x.max()).days),
    Invoice=('invoice_no', 'nunique'),
    Monetary=('total_amount', 'sum')
).reset_index()

rfm.columns = ['CustomerID', 'Recency', 'Invoice', 'Monetary']
rfm = rfm.reset_index(drop=True)

rfm
```

# RFM Segmentation

---
**Please remove this insturction cell after you are done with coding**
RFM segmentation categorizes your customers into different segments, according to their interactions with your website, which will allow you to subsequently approach these groups in the most effective way. In this article, we will show you how to make an RFM segmentation based on an RFM score combining all three RFM parameters together and allowing you to divide your customers into 11 different segments. 

- [RFM Segmentation business cases](https://docs.exponea.com/docs/rfm-segmentation-business-use)

- [RFM Segmentation Guide](https://docs.exponea.com/docs/rfm-segmentation-business-use)

As you can see, computing RFM segmentation requires extensive domain knowledge in marketing which is out of the scope in this project. In practice, you will work with BA/DA to figure out how to compute RFM segments. To simplify this project, a [sample RFM segmentation Notebook](https://github.com/jarviscanada/jarvis_data_eng_demo/blob/feature/data/python_data_wrangling/ipynb/customer-segmentation-with-rfm-score.ipynb) is provided. You are responsible to understand everything from that Notebook and then integrate it into yours. 

- Download the [sample notebook](https://github.com/jarviscanada/jarvis_data_eng_demo/blob/feature/data/python_data_wrangling/ipynb/customer-segmentation-with-rfm-score.ipynb) and import to your Jupyter Notebook or VSCode
- Run the notebook and understand all cells
- Read the remark section at the end of the notebook. You will need this information when writing the README file
- Integrate the RFM segmentation calculation into your notebook

---

```python
# Score each dimension into quartiles (1 = worst, 4 = best)
rfm['R'] = pd.qcut(rfm['Recency'], 4, labels=[4, 3, 2, 1])
rfm['F'] = pd.qcut(rfm['Invoice'].rank(method='first'), 4, labels=[1, 2, 3, 4])
rfm['M'] = pd.qcut(rfm['Monetary'], 4, labels=[1, 2, 3, 4])

# Combine into RFM score string
rfm['RFM_Score'] = rfm['R'].astype(str) + rfm['F'].astype(str) + rfm['M'].astype(str)
rfm['RF_Score'] = rfm['R'].astype(str) + rfm['F'].astype(str)

# Segment mapping
seg_map = {
    r'[1-2][1-2]': 'Hibernating',
    r'[1-2][3-4]': 'At Risk',
    r'3[1-2]':     'About to Sleep',
    r'33':         'Need Attention',
    r'[3-4][4]':   'Loyal Customers',
    r'41':         'Promising',
    r'[4][2-3]':   'Potential Loyalists',
}

rfm['Segment'] = rfm['RF_Score'].replace(seg_map, regex=True)

rfm.head(10)
```

```python
# Plot segment distribution
segment_counts = rfm['Segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Count']

plt.figure(figsize=(12, 6))
sns.barplot(data=segment_counts, x='Segment', y='Count', palette='coolwarm')
plt.title('Customer Segments (RFM)')
plt.xlabel('Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Average RFM values per segment
rfm.groupby('Segment')[['Recency', 'Invoice', 'Monetary']].mean().round(2).sort_values('Monetary', ascending=False)
```
